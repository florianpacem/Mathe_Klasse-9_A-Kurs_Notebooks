# Quadratische Funktionen - Die drei Darstellungsformen

## Einführung

In diesem Notebook lernst du die drei verschiedenen Darstellungsformen einer quadratischen Funktion kennen:

1. **Scheitelpunktform**: $f(x)=a\cdot(x−d)^2+e$
2. **Faktorisierte Form** - auch Nullstellenform genannt: $f(x)=a\cdot(x−x1)\cdot(x−x2)$
3. **Polynomform** - auch allgemeine Form genannt: $f(x)=ax^2+bx+c$

Jede Form hat ihre eigenen Stärken und zeigt unterschiedliche Eigenschaften der Parabel!

Am Ende diesen Kapitels solltest du folgendes gelernt haben:

1. Ich kann die drei Darstellungsarten quadratischer Funktionen aufschreiben.
2. Ich kann die Bedeutung der Parameter in jeder Darstellungsform erklären.
2. Ich kann die Eigenschaften einer Parabel aus jeder Darstellungsform ablesen.
3. Ich kann den Scheitelpunkt aus jeder Darstellungsform bestimmen.

## 1. Die Scheitelpunktform

$$f(x)=a\cdot(x−d)^2+e$$

**Aufgabe:** Verändere die Funktionsgleichung durch bewegen der Slider. Welches Muster erkennst du in der Funktionsgleichung?


```python
def plot_scheitelpunkt(a=1, d=0, e=0):
    x = np.linspace(-10, 10, 400)
    y = a * (x - d)**2 + e
    
    plt.figure(figsize=(10, 6))
    plt.plot(x, y, 'b-', linewidth=2, label=f'$f(x) = {a}(x-{d})^2 + {e}$')
    plt.plot(d, e, 'ro', markersize=10, label=f'Scheitelpunkt S({d}|{e})')
    plt.plot(x, y, 'b-', linewidth=2, label=f'Öffnung: {"nach oben" if a > 0 else "nach unten"}')
    plt.plot(x, y, 'b-', linewidth=2, label=f'Streckung: {"gestreckt" if abs(a) > 1 else ("gestaucht" if abs(a) < 1 else "keine Streckung")}')
    
    plt.axhline(y=0, color='k', linewidth=0.5)
    plt.axvline(x=0, color='k', linewidth=0.5)
    plt.grid(True, alpha=0.3)
    plt.xlim(-10, 10)
    plt.ylim(-10, 10)
    plt.xticks(range(-10, 11, 1))
    plt.yticks(range(-10, 11, 1))
    plt.xlabel('x')
    plt.ylabel('f(x)')
    plt.legend()
    plt.title('Scheitelpunktform')
    plt.show()

interact(plot_scheitelpunkt, 
         a=FloatSlider(min=-3, max=3, step=0.1, value=1, description='a'),
         d=FloatSlider(min=-5, max=5, step=0.5, value=0, description='d'),
         e=FloatSlider(min=-5, max=5, step=0.5, value=0, description='e'))
```

```python
from ipywidgets import Accordion, Output
from IPython.display import display, Markdown

# Frage anzeigen
display(Markdown("**Aufgabe:** Was kann man aus der **Scheitelpunktform** direkt ablesen?"))

# Output-Widget für die Lösung
loesung_output = Output()
with loesung_output:
    display(Markdown(r"""
**Lösung:**

- **Scheitelpunkt**: $S(d|e)$
- **Öffnung**:
    - $a>0$ → nach oben
    - $a<0$ → nach unten
- **Streckung**: 
    - $|a|>1$ → gestreckt
    - $|a|<1$ → gestaucht
    """))

# Akkordeon erstellen
accordion = Accordion(children=[loesung_output])
accordion.set_title(0, '✅ Lösung anzeigen')
accordion.selected_index = None  # Zugeklappt

display(accordion)
```

## 2. Die faktorisierte Form (Nullstellenform)

$$f(x) = a\cdot(x-x_1)\cdot(x-x_2)$$


### 2.1 Wiederholung des Begriffs **Nullstelle**


```python
import numpy as np
import matplotlib.pyplot as plt

# x-Werte
x = np.linspace(-5, 5, 400)

# Drei Parabeln definieren
# 1. Zwei einfache Nullstellen: f(x) = (x+3)(x-2) = x² + x - 6
y1 = (x + 3) * (x - 2)

# 2. Doppelte Nullstelle: f(x) = (x-1)² = x² - 2x + 1
y2 = (x - 1)**2

# 3. Keine Nullstellen: f(x) = x² - 4x + 6
y3 = x**2 - 4*x + 6

# Plot erstellen
plt.figure(figsize=(12, 8))

# Drei Parabeln zeichnen
plt.plot(x, y1, 'b-', linewidth=2.5, label='Zwei Nullstellen: $f(x) = (x+3)(x-2)$')
plt.plot(x, y2, 'g-', linewidth=2.5, label='Eine Nullstelle: $f(x) = (x-1)^2$')
plt.plot(x, y3, 'r-', linewidth=2.5, label='Keine Nullstellen: $f(x) = x^2 - 4x + 6$')

# Nullstellen markieren
# Parabel 1: Nullstellen bei x = -3 und x = 2
plt.plot([-3, 2], [0, 0], 'bo', markersize=10)

# Parabel 2: Doppelte Nullstelle bei x = 1
plt.plot(1, 0, 'go', markersize=10)

# Koordinatensystem
plt.axhline(y=0, color='k', linewidth=0.8)
plt.axvline(x=0, color='k', linewidth=0.8)
plt.grid(True, alpha=0.3)

# Achsenbeschriftung
plt.xlim(-5, 5)
plt.ylim(-7, 10)
plt.xticks(range(-5, 6, 1))
plt.yticks(range(-7, 11, 1))
plt.xlabel('x', fontsize=12)
plt.ylabel('f(x)', fontsize=12)

# Legende und Titel
plt.legend(fontsize=11, loc='upper left')
plt.title('Nullstellen quadratischer Funktionen', fontsize=14, fontweight='bold')

plt.show()
```

### 2.2 Ablesbare Eigenschaften der Parabel

**Aufgabe:** Verändere die Funktionsgleichung durch Bewegen der Slider. Welches Muster erkennst du in der Funktionsgleichung?

```python
def plot_faktorisiert(a=1, x1=-2, x2=2):
    x = np.linspace(-10, 10, 400)
    y = a * (x - x1) * (x - x2)
    
    # Scheitelpunkt berechnen
    d = (x1 + x2) / 2
    e = a * (d - x1) * (d - x2)
    
    plt.figure(figsize=(6, 6))
    plt.plot(x, y, 'g-', linewidth=2, label=f'$f(x) = {a} · (x-{x1}) · (x-{x2})$')
    plt.plot([x1, x2], [0, 0], 'ro', markersize=10, label=f'Nullstellen: x₁={x1}, x₂={x2}')
    plt.plot(x, y, 'g-', linewidth=2, label=f'Öffnung: {"nach oben" if a > 0 else "nach unten"}')
    plt.plot(x, y, 'g-', linewidth=2, label=f'Streckung: {"gestreckt" if abs(a) > 1 else ("gestaucht" if abs(a) < 1 else "keine Streckung")}')
    
    plt.xticks(range(-10, 11, 1))  # x-Achse in 1er-Schritten
    plt.yticks(range(-10, 11, 1))  # y-Achse in 1er-Schritten
    plt.axhline(y=0, color='k', linewidth=0.5)
    plt.axvline(x=0, color='k', linewidth=0.5)
    plt.grid(True, alpha=0.3)
    plt.xlim(-10, 10)
    plt.ylim(-10, 10)
    plt.xlabel('x')
    plt.ylabel('fxx')
    plt.legend()
    plt.title('Faktorisierte Form - Nullstellenform')
    plt.show()

interact(plot_faktorisiert,
         a=FloatSlider(min=-3, max=3, step=0.1, value=1, description='a'),
         x1=FloatSlider(min=-8, max=8, step=0.5, value=-2, description='x₁'),
         x2=FloatSlider(min=-8, max=8, step=0.5, value=2, description='x₂'))
```

#### Zusammenfassung: Eigenschaften der Parabel ausgehend von der Nullstellenform

An einer Funktionsgleichung, die in Nullstellenform gegeben ist,

$$f(x) = a \dot (x - x_{1}) \cdot (x - x_{2})$$

können wir folgendes direkt ablesen:

- **Nullstellen**: $x_1$ und $x_2$ (*oder evtl. nur eine Nullstelle*)
- **Öffnung**:
    - $a>0$ → nach oben
    - $a<0$ → nach unten
- **Streckung**: 
    - $|a|>1$ → gestreckt
    - $|a|<1$ → gestaucht

### 2.3 Bestimmung des Scheitelpunkts ausgehend von der Nullstellenform

```python
from ipywidgets import Accordion, Output
from IPython.display import display, Markdown

# Aufgabe
display(Markdown("**Frage:** Wo liegt die **x-Koordinate des Scheitelpunkts** in Bezug auf die Nullstellen?"))

# Akkordeon 1: Hilfestellung
hilfe_output = Output()
with hilfe_output:
    display(Markdown(r"""

**Hilfestellung**

Bewege die Slider. Achte auf die Lage der Nullstellen und die Lage des Scheitelpunkts.
    """))

    def plot_faktorisiert(a=1, x1=-2, x2=2):
        x = np.linspace(-10, 10, 400)
        y = a * (x - x1) * (x - x2)
        
        # Scheitelpunkt berechnen
        d = (x1 + x2) / 2
        e = a * (d - x1) * (d - x2)
        
        plt.figure(figsize=(10, 6))
        plt.plot(x, y, 'g-', linewidth=2, label=f'$f(x) = {a}(x-{x1})(x-{x2})$')
        plt.plot([x1, x2], [0, 0], 'ro', markersize=10, label=f'Nullstellen: $x_1={x1}$, $x_2={x2}$')
        plt.plot(d, e, 'mo', markersize=10, label=f'Scheitelpunkt S({d:.1f}|{e:.1f})')
        plt.axhline(y=0, color='k', linewidth=0.5)
        plt.axvline(x=0, color='k', linewidth=0.5)
        plt.grid(True, alpha=0.3)
        plt.xlim(-10, 10)
        plt.ylim(-10, 10)
        plt.xticks(range(-10, 11, 1))
        plt.yticks(range(-10, 11, 1))
        plt.xlabel('x')
        plt.ylabel('f(x)')
        plt.legend()
        plt.title('Faktorisierte Form (Nullstellenform)')
        plt.show()

    interact(plot_faktorisiert,
         a=FloatSlider(min=-3, max=3, step=0.1, value=1, description='a'),
         x1=FloatSlider(min=-8, max=8, step=0.5, value=-2, description='x₁'),
         x2=FloatSlider(min=-8, max=8, step=0.5, value=2, description='x₂'))

hilfe_accordion = Accordion(children=[hilfe_output])
hilfe_accordion.set_title(0, '💡 Hilfestellung anzeigen')
hilfe_accordion.selected_index = None

display(hilfe_accordion)

# Akkordeon 2: Lösung
loesung_output = Output()
with loesung_output:
    display(Markdown(r"""

**Lösung:**

Die x-Koordinate des Scheitelpunkts $d$ liegt immer **in der Mitte der beiden Nullstellen**.
    """))

loesung_accordion = Accordion(children=[loesung_output])
loesung_accordion.set_title(0, '✅ Lösung anzeigen')
loesung_accordion.selected_index = None
display(loesung_accordion)
```

```python
from ipywidgets import Accordion, Output
from IPython.display import display, Markdown

# Frage anzeigen
display(Markdown("**Frage:** Angenommen wir kennen die x-Koordinate $d$ des Scheitelpunkts. Wie bestimmen wir dann die **y-Koordinate $e$ des Scheitelpunkts**?"))

# Output-Widget für die Lösung
loesung_output = Output()
with loesung_output:
    display(Markdown(r"""
**Lösung:**

Immer wenn wir zu einem gegebenen x-Wert den dazugehörigen y-Wert bestimmen möchten, gehen wir wie folgt vor:

1. Wir setzen den x-Wert (in diesem Fall wäre das $d$) in die Funktionsgleichung ein.
2. Wir rechnen alles aus. Das Ergebnis ist der dazugehörige y-Wert (in diesem Fall wäre das $e$).
    """))

# Akkordeon erstellen
accordion = Accordion(children=[loesung_output])
accordion.set_title(0, '✅ Lösung anzeigen')
accordion.selected_index = None  # Zugeklappt

display(accordion)
```

#### Zusammenfassung: Bestimmung des Scheitelpunkts ausgehend von der Nullstellenform

Um den Scheitelpunkt $S(d|e)$ ausgehend von einer Funktionsgleichung in Nullstellenform zu bestimmen, gehen wir wie folgt vor:

**1. Wir bestimmen die x-Koordinate $d$ des Scheitelpunkts**

Da der Scheitelpunkt immer in der Mitte zwischen den beiden Nullstellen liegt, können wir seine x-Koordinate folgendermaßen berechnen:

$$d = \frac{x_{1} + x_{2}}{2}$$

**2. Wir bestimmen die y-Koordinate $e$ des Scheitelpunkts**

Wir setzen $d$ in die Funktionsgleichung ein und berechnen dadurch $e$.

$$e = f(d)$$

## 3. Die Polynomform (allgemeine Form)

$$f(x) = ax^2 + bx + c$$


### 3.1 Wiederholung des Begriffs **y-Achsenabschnitt**



```python
import numpy as np
import matplotlib.pyplot as plt

# x-Werte
x = np.linspace(-6, 6, 400)

# Drei Parabeln definieren
# 1. Linker Ast schneidet y-Achse: Scheitelpunkt rechts von y-Achse
#    f(x) = (x-3)² - 2 (nach oben geöffnet)
y1 = (x - 3)**2 - 2

# 2. Rechter Ast schneidet y-Achse: Scheitelpunkt links von y-Achse, nach unten geöffnet
#    f(x) = -(x+2)² + 5
y2 = -(x + 2)**2 + 5

# 3. Scheitelpunkt auf y-Achse: Scheitelpunkt bei x=0
#    f(x) = -0.5(x)² + 4
y3 = -0.5 * x**2 + 4

# Y-Achsenabschnitte berechnen
c1 = (0 - 3)**2 - 2  # = 7
c2 = -(0 + 2)**2 + 5  # = 1
c3 = 4

# Plot erstellen
plt.figure(figsize=(12, 8))

# Drei Parabeln zeichnen
plt.plot(x, y1, 'b-', linewidth=2.5, label=f'Linker Ast: $f(x) = (x-3)^2 - 2$ (c = {c1})')
plt.plot(x, y2, 'g-', linewidth=2.5, label=f'Rechter Ast: $f(x) = -(x+2)^2 + 5$ (c = {c2})')
plt.plot(x, y3, 'r-', linewidth=2.5, label=f'Scheitelpunkt: $f(x) = -0.5x^2 + 4$ (c = {c3})')

# Y-Achsenabschnitte markieren
plt.plot(0, c1, 'bo', markersize=12, zorder=5)
plt.plot(0, c2, 'go', markersize=12, zorder=5)
plt.plot(0, c3, 'ro', markersize=12, zorder=5)

# Koordinatensystem
plt.axhline(y=0, color='k', linewidth=0.8)
plt.axvline(x=0, color='k', linewidth=1.2)  # y-Achse etwas dicker
plt.grid(True, alpha=0.3)

# Achsenbeschriftung
plt.xlim(-6, 6)
plt.ylim(-4, 8)
plt.xticks(range(-6, 7, 1))
plt.yticks(range(-4, 9, 1))
plt.xlabel('x', fontsize=12)
plt.ylabel('f(x)', fontsize=12)

# Legende und Titel
plt.legend(fontsize=11, loc='upper right')
plt.title('y-Achsenabschnitt bei quadratischen Funktionen', fontsize=14, fontweight='bold')

plt.show()
```

### *Hier fehlt noch eine entsprechende Aufgabe*

```python
def plot_polynom(a=1, b=0, c=0):
    x = np.linspace(-10, 10, 400)
    y = a * x**2 + b * x + c
    
    # Scheitelpunkt berechnen
    d = -b / (2*a)
    e = a * d**2 + b * d + c
    
    # Nullstellen berechnen (falls vorhanden)
    discriminant = b**2 - 4*a*c
    
    plt.figure(figsize=(6, 6))
    plt.plot(x, y, 'r-', linewidth=2, label=f'$f(x) = {a}x^2 + {b}x + {c}$')
    plt.plot(0, c, 'bo', markersize=10, label=f'y-Achsenabschnitt: c={c}')
    plt.plot(d, e, 'mo', markersize=10, label=f'Scheitelpunkt S({d:.1f}|{e:.1f})')
    plt.xticks(range(-10, 11, 1))  # x-Achse in 1er-Schritten
    plt.yticks(range(-10, 11, 1))  # y-Achse in 1er-Schritten
    
    if discriminant >= 0:
        x1 = (-b + np.sqrt(discriminant)) / (2*a)
        x2 = (-b - np.sqrt(discriminant)) / (2*a)
        plt.plot([x1, x2], [0, 0], 'ro', markersize=10, label=f'Nullstellen: {x1:.1f}, {x2:.1f}')
    
    plt.axhline(y=0, color='k', linewidth=0.5)
    plt.axvline(x=0, color='k', linewidth=0.5)
    plt.grid(True, alpha=0.3)
    plt.xlim(-10, 10)
    plt.ylim(-10, 10)
    plt.xlabel('x')
    plt.ylabel('f(x)')
    plt.legend()
    plt.title('Polynomform (Normalform)')
    plt.show()
    
    print(f"Y-Achsenabschnitt: c = {c}")
    print(f"Scheitelpunkt (berechnet): S({d:.2f}|{e:.2f})")
    if discriminant >= 0:
        print(f"Nullstellen (berechnet): x₁ = {x1:.2f}, x₂ = {x2:.2f}")
    else:
        print("Keine reellen Nullstellen")

interact(plot_polynom,
         a=FloatSlider(min=-3, max=3, step=0.1, value=1, description='a'),
         b=FloatSlider(min=-10, max=10, step=0.5, value=0, description='b'),
         c=FloatSlider(min=-10, max=10, step=0.5, value=0, description='c'))
```

| Darstellungsform | Direkt ablesbar | Beste Anwendung |
|------------------|-----------------|-----------------|
| **Scheitelpunktform** $a(x-d)^2 + e$ | Scheitelpunkt $S(d|e)$ | Graph zeichnen, Verschiebungen verstehen |
| **Faktorisierte Form** $a(x-x_1)(x-x_2)$ | Nullstellen $x_1, x_2$ | Schnittpunkte mit x-Achse finden |
| **Polynomform** $ax^2 + bx + c$ | y-Achsenabschnitt $c$ | Funktionswerte berechnen, Ableitungen, ... |
